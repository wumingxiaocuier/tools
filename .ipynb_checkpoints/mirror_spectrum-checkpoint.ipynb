```python
"""
绘制镜像谱：
    给定谱图(peaks)、peptide(seq+mod)、pPred的ms2预测结果，绘制镜像谱
"""
import matplotlib.pyplot as plt
import numpy as np
from utils import *
import math
```

```python
element2mass = get_element2mass("infos/element.ini")
aa2mass = get_aa2mass("infos/aa.ini", element2mass)
mod2mass = get_mod2mass("infos/modification.ini")

def cal_matched_peaks(peaks, modified_seq, precursor_charge, topk=None, ppm=20):
    """
    使用方法:
        cal_by_matches(peaks, "PANITDSLSNR", 3, topk=200)
    """
    if topk is not None:
        intensities = [peak[1] for peak in peaks]
        intensity_th = quickselect(intensities, topk)
        if intensity_th != -1:
            _peaks = list()
            for mz, intensity in peaks:
                if intensity >= intensity_th:
                    _peaks.append([mz, intensity])
            peaks = _peaks
            
    def search_mz(peaks, target_mz, ppm=20):
        """
        如果找到匹配峰，则返回该峰的mz, intensity, error, 否则返回-1, -1, -1
        """
        left_mz, right_mz = (1 - ppm * 1e-6) * target_mz, (1 + ppm * 1e-6) * target_mz
        left, right = 0, len(peaks)  # 左闭右开区间
        while left < right:
            mid = (left + right) // 2
            if left_mz <= peaks[mid][0] <= right_mz:
                return mid
            elif peaks[mid][0] < left_mz:
                left = mid + 1
            else:
                right = mid
        return -1

    seq, mods = modified_seq.split("\t")
    n = len(seq)

    # 计算每个位点处，氨基酸+修饰的质量
    r_mass = [0.0] * n
    for i in range(n):
        r_mass[i] = aa2mass[seq[i]]
    for mod in mods.strip(";").split(";"):
        mod_pos,mod_name = mod.split(",")
        mod_pos = int(mod_pos)
        if mod_pos==0:  # N-term
            r_mass[0] += mod2mass[mod_name]
        else:  # normal & C-term
            r_mass[mod_pos-1] += mod2mass[mod_name] 

    matched_peaks = list()  # 匹配情况
    b_mz = 0.0
    for i in range(n - 1):
        b_mz += r_mass[i]
        for charge in range(1,precursor_charge+1):
            matched_id = search_mz(peaks, b_mz/charge+P_mass, ppm=ppm)
            print((f"b{i+1}"+(charge*'+'), b_mz/charge+P_mass))
            if matched_id >= 0:  # 有匹配峰
                matched_peaks.append((f"b{i+1}"+(charge*'+'), *peaks[matched_id]))
    y_mz = H2O_mass
    for i in range(n - 1, 0, -1):
        y_mz += r_mass[i]
        for charge in range(1,precursor_charge+1):
            matched_id = search_mz(peaks, y_mz/charge+P_mass, ppm=ppm)
            print((f"y{n-i}"+(charge*'+'), y_mz/charge+P_mass))
            if matched_id >= 0:
                matched_peaks.append((f"y{n-i}"+(charge*'+'), *peaks[matched_id]))

    return matched_peaks


def cal_pred_peaks(pred_ms2, modified_seq, precursor_charge):
    seq, mods = modified_seq.split("\t")
    n = len(seq)

    # 计算每个位点处，氨基酸+修饰的质量
    r_mass = [0.0] * n
    for i in range(n):
        r_mass[i] = aa2mass[seq[i]]
    for mod in mods.strip(";").split(";"):
        mod_pos,mod_name = mod.split(",")
        mod_pos = int(mod_pos)
        if mod_pos==0:  # N-term
            r_mass[0] += mod2mass[mod_name]
        else:  # normal & C-term
            r_mass[mod_pos-1] += mod2mass[mod_name] 

    mzs_record = list()
    b_mz = 0.0
    for i in range(n - 1):  # b1-b(n-1)
        b_mz += r_mass[i]
        for charge in range(1,precursor_charge+1):
            mzs_record.append(b_mz/charge+P_mass)
    y_mz = H2O_mass
    for i in range(n - 1, 0, -1):  # y1-y(n-1)
        y_mz += r_mass[i]
        for charge in range(1,precursor_charge+1):
            mzs_record.append(y_mz/charge+P_mass)

    pred_peaks = list()
    for ion_type, pos, inten in pred_ms2:
        # 计算它在mzs_record中的下标: (ion_type,pos)->ion_name->idx
        charge = ion_type//2+1
        is_b = True if (ion_type%2==0) else False
        num = pos+1 if is_b else n-pos-1
        ion_name = f"b{num}{'+'*charge}" if is_b else f"y{num}{'+'*charge}"
        idx = (num-1)*precursor_charge+(charge-1) if is_b else (n+num-2)*precursor_charge+(charge-1)
        pred_peaks.append((ion_name, mzs_record[idx], inten))

    return pred_peaks


def cal_cos(matched_peaks, pred_peaks):
    """
    计算实验谱图与预测谱图的cosine相似度
    [(ion_name, m/z, intensity)......]
    """
    p1, p2 = 0, 0
    sim = 0.0
    matched_peaks = sorted(matched_peaks, key=lambda item:item[0])
    pred_peaks = sorted(pred_peaks, key=lambda item:item[0])
    module1 = sum([inten*inten for _,_,inten in matched_peaks])
    module2 = sum([inten*inten for _,_,inten in pred_peaks])
    while p1<len(matched_peaks) and p2<len(pred_peaks):
        if matched_peaks[p1][0] == pred_peaks[p2][0]:
            sim += matched_peaks[p1][2]*pred_peaks[p2][2]
            p1 += 1
            p2 += 1
        elif matched_peaks[p1][0] < pred_peaks[p2][0]:
            p1 += 1
        else:
            p2 += 1
    # print(sim, module1, module2)
    return sim / math.sqrt(module1*module2)
```

```python
def draw_mirrored_spectrum(peaks, pred_ms2, modified_seq, charge):
    """
    绘制镜像谱图：
        peaks: [(m/z,intensity),...]
        pred_ms2(对pPred的输出稍微转换一下): [(ion_type,pos,relative_intensity),...]
        modified_seq: seq\tmodifications    
    """
    # 计算匹配峰 matched_peaks: [(ion_name,m/z,intensity),......]
    matched_peaks = cal_matched_peaks(peaks, modified_seq, charge)
    # 根据预测 ms2 计算匹配谱峰 pred_ms2: [(ion_name,m/z,intensity),...]
    pred_peaks = cal_pred_peaks(pred_ms2, modified_seq, charge)
    # 计算实验谱图与预测谱图的cosine相似度
    cos = cal_cos(matched_peaks, pred_peaks)
    
    # 创建图形
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.set_title("Mirrored Spectrum", fontsize=14)
    ax.set_xlabel("m/z", fontsize=12)
    ax.set_ylabel("Normalized Intensity", fontsize=12)
    
    # 绘制所有谱峰(上方)
    flag = False
    for mz, inten in peaks:
        ax.vlines(mz, 0, inten, color="black", lw=1)  # 蓝色线条
    # 绘制匹配谱峰(上方)
    flag = False
    for ion_name, mz, inten in matched_peaks:
        if not flag:
            ax.vlines(mz, 0, inten, color="orange", lw=2, label="Experimental")  # 蓝色线条
            flag = True
        else:
            ax.vlines(mz, 0, inten, color="orange", lw=2)  # 蓝色线条
        ax.text(mz, inten, ion_name, 
                ha='center', va='bottom', fontsize=8, rotation=45, color="orange")  # 上方标注
    flag = False
    # 绘制预测谱图(下方)
    for ion_name, mz, inten in pred_peaks:
        if not flag:
            ax.vlines(mz, 0, -inten, color="#d62728", lw=2, label="Predicted")  # 红色线条
            flag = True
        else:
            ax.vlines(mz, 0, -inten, color="#d62728", lw=2)  # 红色线条
        ax.text(mz, -inten, ion_name, 
                ha='center', va='top', fontsize=8, rotation=45, color="#d62728")  # 下方标注

    ax.text(0.5, 1.05, f"Cosine Similarity: {cos:.4f}", 
        horizontalalignment='center', 
        verticalalignment='bottom', 
        transform=ax.transAxes,  # 使用轴坐标系统
        fontsize=12, 
        fontweight='bold',
        color='purple')
    ax.text(0.5, 0.75, modified_seq.split('\t')[0],
            transform=ax.transAxes,  # 使用轴坐标系统
            fontsize=15, 
            # fontweight='bold',
           )
    # 添加水平轴
    ax.axhline(0, color='black', lw=1.5)
    # 美化设置
    ax.grid(True, linestyle='--', alpha=0.7, axis='x')
    ax.tick_params(axis='both', which='major', labelsize=10)
    ax.legend(loc='upper right', fontsize=10)
    plt.tight_layout()
    plt.show()
```

```python
"""
给定一张谱图、谱图对应的候选肽段、以及pPred对这些肽段的预测结果，生成镜像谱图
"""
# 获取谱峰信息
mgf_path = "data/draw_mirrored_spectrum/test.mgf"
spec_name = "20180425_QE_HFX_LC2_HLAIp_CHC_SA_normC3N02671_10_R3.14229.14229.2.2.dta"
peaks = list()
fw = open(mgf_path,"r")
flag = False
while True:
    line = fw.readline()
    if not line:
        break
    if line.startswith("T"):
        name = line.strip()[6:]
        if name == spec_name:
            flag = True
    elif line.startswith('C'):
        charge = int(line.strip().strip('+').split('=')[1])
    elif 'A' <= line[0] <= 'Z':
        continue
    elif flag:
        while not line.startswith('E'):
            items = line.strip().split()
            mz,inten = float(items[0]),float(items[1])
            peaks.append([mz,inten])
            line = fw.readline()
        break
fw.close()
max_inten = -1
for mz,inten in peaks:
    max_inten = max(max_inten, inten)
for i in range(len(peaks)):
    peaks[i][1] /= max_inten
        
# 候选肽段列表
pep_path = "data/draw_mirrored_spectrum/test.pepdata"
modified_seqs = list()
fw = open(pep_path, "r")
next(fw,None)
for line in fw:
    items = line.strip().split('\t')
    modified_seqs.append(items[0]+'\t'+items[1])
fw.close()

# 预测结果 (与pep_path中的肽段一一对应)
pred_path = "data/draw_mirrored_spectrum/test.pred.ms2data"
pred_ms2 = list()
fw = open(pred_path,"r")
next(fw,None)
for line in fw:
    pred_items = line.strip().split("\t")
    start_pos = charge*2
    p_ms2 = list()
    for i in range(int(pred_items[start_pos-1])):
        ion_type,pos,intensity = int(pred_items[start_pos+3*i]),int(pred_items[start_pos+3*i+1]),float(pred_items[start_pos+3*i+2])
        p_ms2.append((ion_type,pos,intensity))
    pred_ms2.append(p_ms2)
fw.close()
```
