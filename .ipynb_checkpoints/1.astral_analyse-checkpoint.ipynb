## Analyze

    idxspfind_res: idx->[[seq,modified_seq,charge,is_decoy,cal_mh,exp_mh,q_value,protein_names]]

    idx2diann_res: idx->[[seq,modified_seq,charge,is_decoy,cal_mz,q_value,protein_ids_str]]

    注意：idx, seq, modified_seq, is_decoy已对齐

    pfind_res包含全部结果(不卡FDR),DIA-NN仅包含FDR阈值下的结果

    根据pfind.spectra的特点，显然某个idx对应的肽段结果(混合谱)都是按q_value从小到大排序好的

    注意：以下对于pFind来说，将肽段分为鉴定到的+未通过阈值+没有结果的

```python
"""
获取肽段对应信息：
pep->idxs
seq->info[is_decoy,protein_names]: 这里有个很奇怪的问题——DIA-NN和pFind的seq->pro_names无法对齐，已经找到了其中三个原因：
    pFind推断蛋白质时会对I/L进行替换，找到所有可能的蛋白质；pFind有REV蛋白质；DIA-NN会少算一部分正常蛋白质
pfind_peps, diann_peps, intersect_peps......
"""
# seq -> [is_decoy,protein_names]  这些属性对于确定的肽段来说是固定的
pfind_seq2info = dict()


# 通过阈值的肽段序列、修饰肽段
pfind_seqs = set()
pfind_modified_seqs = set()
# 未通过阈值的肽段序列、修饰肽段
pfind_seqs_bad = set()
pfind_modified_seqs_bad = set()
# pFind所有(有结果的)修饰肽段、肽段序列向pfind_idx的映射: modified_seq -> [pfind_idx1,...]
pfind_modified_seq2idxs = dict()
pfind_seq2idxs = dict()


for pfind_idx, infos in idx2pfind_res.items():
    for info in infos:  # 混合谱
        seq,modified_seq,charge,is_decoy,cal_mh,exp_mh,q_value,protein_names = info
        
        # 修饰肽段
        if q_value <= pep_level_fdr:
            pfind_modified_seqs.add(modified_seq)
        else:
            pfind_modified_seqs_bad.add(modified_seq)
        if modified_seq not in pfind_modified_seq2idxs:
            pfind_modified_seq2idxs[modified_seq] = [pfind_idx]
        else:
            pfind_modified_seq2idxs[modified_seq].append(pfind_idx)
        # 肽段序列
        if q_value <= pep_level_fdr:
            pfind_seqs.add(seq)
        else:
            pfind_seqs_bad.add(seq)
        if seq not in pfind_seq2idxs:
            pfind_seq2idxs[seq] = [pfind_idx]
        else:
            pfind_seq2idxs[seq].append(pfind_idx)
            
        if seq not in pfind_seq2info:
            pfind_seq2info[seq] = [is_decoy,protein_names]

    
diann_seq2info = dict()
# seq -> idxs
diann_seqs = set()
diann_modified_seqs = set()
# modified_seq -> idxs
diann_seq2idxs = dict()
diann_modified_seq2idxs = dict()
a = 0
for pfind_idx, infos in idx2diann_res.items():
    a += 1
    if a % 1000 == 0:
        print(a)
    for info in infos:  # 混合谱
        seq,modified_seq,charge,is_decoy,cal_mz,q_value,protein_ids_str = info
        # 修饰肽段
        if modified_seq not in diann_modified_seq2idxs:
            diann_modified_seq2idxs[modified_seq] = [pfind_idx]
            diann_modified_seqs.add(modified_seq)
        else:
            diann_modified_seq2idxs[modified_seq].append(pfind_idx)
        # 肽段序列
        if seq not in diann_seq2idxs:
            diann_seq2idxs[seq] = [pfind_idx]
            diann_seqs.add(seq)
        else:
            diann_seq2idxs[seq].append(pfind_idx)

        if seq not in diann_seq2info:  # 按pFind标准来
            # protein_names = parse(protein_ids_str)
            protein_names = get_seq2pro(seq, pro2seq, replace=True)
            diann_seq2info[seq] = [is_decoy,protein_names]
            
            
print(f"pFind鉴定到的肽段序列数目: {len(pfind_seqs)}")
print(f"pFind鉴定到的修饰肽段数目: {len(pfind_modified_seqs)}")
print(f"pFind未通过阈值的肽段序列数目: {len(pfind_seqs_bad)}")
print(f"pFind未通过阈值的修饰肽段数目: {len(pfind_modified_seqs_bad)}")
print(f"DIA-NN鉴定到的肽段序列数目: {len(diann_seqs)}")
print(f"DIA-NN鉴定到的修饰肽段数目: {len(diann_modified_seqs)}")
del pfind_modified_seqs
del pfind_modified_seqs_bad
del pfind_modified_seq2idxs
del diann_modified_seqs
del diann_modified_seq2idxs
```

## 考察DIA-NN结果的正确性

```python
"""
考察pFind & DIA-NN鉴定肽段的正确率：陷阱库方法
    蛋白质搜索文件中包含human、拟南芥(均以"_ARATH"结尾)和常见污染蛋白质，ARATH作为陷阱库
    若鉴定到的肽段对应的蛋白质列表中，除开decoy蛋白质外仅包含ARATH蛋白质，则其落入陷阱库
"""
trap_tag = "_ARATH"
diann_trap_num = 0
for diann_seq in diann_seqs:  # DIA-NN鉴定到的target肽段
    seq_info = diann_seq2info[diann_seq]
    protein_names = seq_info[1]
    
    flag = True
    for pro in protein_names:
        if not pro.endswith(trap_tag) and not pro.startswith("REV_"):
            flag = False
            break

    # flag = False
    # for pro in protein_names:
    #     if pro.endswith(trap_tag):
    #         flag = True
    #         break
    if flag:
        diann_trap_num += 1


pfind_trap_num = 0
for pfind_seq in pfind_seqs:  # pFind鉴定到的肽段
    seq_info = diann_seq2info[pfind_seq]
    is_decoy = seq_info[0]
    if is_decoy:  # 仅考察pFind鉴定到的target肽段
        continue
    protein_names = seq_info[1]
    
    flag = True
    for pro in protein_names:
        if not pro.endswith(trap_tag) and not pro.startswith("REV_"):
            flag = False
            break
    if flag: 
        pfind_trap_num += 1

print(f"DIA-NN鉴定到的{len(diann_seqs)}个肽段序列中，有{diann_trap_num}个落入陷阱库")
print(f"pFind鉴定到的{len(pfind_seqs)}个肽段序列中，有{pfind_trap_num}个落入陷阱库")
```

## 筛选可靠肽段并验证筛选方法的合理性

```python
"""
可信肽段：以不同电荷、修饰形态出现过，或其对应蛋白质不孤立等
"""
# 肽段序列可信度等级(大于0的皆为可信肽段)
seq_conf = dict()

# pFind和DIA-NN鉴定到的肽段集合(并集)
seqs = pfind_seqs | diann_seqs
# pFind和DIA-NN鉴定到的肽段集合(交集)
seqs_intersect = pfind_seqs & diann_seqs

for seq in diann_seqs:
    if seq2info[seq][0] == 1:  # decoy肽段可信度为0
        seq_conf[seq] = 0
        continue
    
    pfind_charge_keys, pfind_mod_keys = set(), set()
    diann_charge_keys, diann_mod_keys = set(), set()

    diann_idxs = diann_seq2idxs[seq] if seq in diann_seq2idxs else []
    for diann_idx in diann_idxs:  # 一个pep可能对应多张谱图
        diann_ress = idx2diann_res[diann_idx]
        for diann_res in diann_ress:  # 可能为混合谱
            if seq == diann_res[0]:
                diann_charge_keys.add(diann_res[2])
                diann_mod_keys.add(diann_res[1])
                
    pfind_idxs = pfind_seq2idxs[seq] if seq in pfind_seq2idxs else []
    for pfind_idx in pfind_idxs:  # 一个pep可能对应多张谱图
        pfind_ress = idx2pfind_res[pfind_idx]
        for pfind_res in pfind_ress:  # 可能为混合谱
            if seq == pfind_res[0] and pfind_res[6] <= pep_level_fdr:
                pfind_charge_keys.add(pfind_res[2])
                pfind_mod_keys.add(pfind_res[1])
                    
    charge_keys = pfind_charge_keys | diann_charge_keys
    mod_keys = pfind_mod_keys | diann_mod_keys

    if len(charge_keys) > 1 and len(mod_keys) > 1:  # 在pFind|DIA-NN结果中以不同电荷、修饰的形式鉴定到：非常可信！
        seq_conf[seq] = 4
        # print(charge_keys, mod_keys)
    elif len(charge_keys) > 1:  # 以不同电荷的形式鉴定到：很可信！
        seq_conf[seq] = 3
    elif len(mod_keys) > 1:  # 以不同修饰的形式鉴定到：可信！
        seq_conf[seq] = 0
    elif len(pfind_charge_keys) >= 1 and len(diann_charge_keys) >= 1:  # pFind和DIA-NN均鉴定到同一形式的该肽段：可信！
        seq_conf[seq] = 0
    else:  # pFind or DIA-NN单独鉴定到：不确定
        seq_conf[seq] = 0


"""
从上一个单元的结果中可以看出，pfind的decoy肽段所属蛋白质大多对应1个支持肽段(少部分对应2个)。
考虑到DIA-NN检测到更多肽段，因此可以把所属蛋白质对应支持肽段大于2的肽段列为可信肽段。
"""
pro2num = dict()
for seq in diann_seqs:
    protein_names = diann_seq2info[seq][1]
    for protein_name in protein_names:
        if protein_name in pro2num:
            pro2num[protein_name] += 1
        else:
            pro2num[protein_name] = 1


threshold = 100000
a = 0
for seq in diann_seqs:
    is_decoy, protein_names = diann_seq2info[seq]
    max_num = 0
    for protein_name in protein_names:
        max_num = max(max_num, pro2num[protein_name])
    if max_num > threshold:  # 如果肽段所属蛋白质有大于threshold个支持肽段，则可信度等级列为1
        seq_conf[seq] = 1
        a += 1
print(f'pFind和DIA-NN鉴定到的肽段(并集)中，所属蛋白质有大于{threshold}个支持肽段的肽段个数: {a}')
```

```python
"""
验证这种可信肽段筛选方法是否合理：
    以包含更少的陷阱库肽段为标准
"""
ori_trap_num = 0
for seq in diann_seqs:
    seq_info = diann_seq2info[seq]
    is_decoy = seq_info[0]
    if is_decoy:  # 仅考察pFind的非decoy肽段
        continue
    protein_names = seq_info[1]
    flag = True
    for pro in protein_names:
        if not pro.endswith(trap_tag) and not pro.startswith("REV_"):
            flag = False
            break
    if flag:
        ori_trap_num += 1
print(f"未经筛选时，DIA-NN结果中包含{ori_trap_num}个陷阱肽段")

cur_trap_num = 0
for seq in diann_seqs:
    seq_info = diann_seq2info[seq]
    is_decoy = seq_info[0]
    if is_decoy or seq_conf[seq] == 0:  # 仅考察pFind的非decoy、且可信度大于0肽段
        continue
    protein_names = seq_info[1]
    flag = True
    for pro in protein_names:
        if not pro.endswith(trap_tag) and not pro.startswith("REV_"):
            flag = False
            break
    if flag:
        print(protein_names, seq_conf[seq], seq, seq2num[seq])
        cur_trap_num += 1

print(f"经筛选后，DIA-NN结果中包含{cur_trap_num}个陷阱肽段")
```

## 查看pFind未鉴定出DIA-NN部分肽段的原因

```python
"""
以DIA-NN鉴定到的肽段序列为中心，分析pFind没有鉴定到的原因
"""
# pFind和DIA-NN共同鉴定到的肽段序列数目 v.s. pFind未鉴定到的肽段序列数目
identify_cnt = 0
not_identify_cnt = 0
# DIA-NN鉴定到，pFind找到但未通过阈值的肽段序列：find_not_pass_cnt = find_not_pass_conf_cnt + find_not_pass_unknown_cnt
find_not_pass_cnt = 0
find_not_pass_conf_cnt = 0
find_not_pass_unknown_cnt = 0
# DIA-NN鉴定而pFind未找到的肽段序列
not_find_cnt = 0
not_find_conf_cnt = 0  # DIA-NN鉴定到的肽段是可信的，但pFind却没鉴定到，这是需要重点关注的肽段
not_find_conf_mass_cnt = 0  # 源于母离子质荷比未检测出
not_find_conf_identify_cnt = 0  # 源于鉴定问题(可能在top10中，但不是top1)
not_find_unknown_cnt = 0  # DIA-NN鉴定到的肽段是正确性未知，此时似乎只能肉眼观察了？
not_find_unknown_mass_cnt = 0  # 源于母离子质荷比未检测出
not_find_unknown_identify_cnt = 0  # 源于鉴定问题(可能在top10中，但不是top1)
# pFind未找到的可信肽段序列对应的谱图: pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, protein_names, problem
focused_spectra = list()  # 可信
focused_spectra_ = list()  # 未知正确性


scan2precursors_info = get_scan2precursors_info(csv_paths)
for diann_seq in diann_seqs:   # 遍历每个DIA-NN鉴定到的肽段序列，DIA-NN中不包括decoy肽段
    """
    是否共同鉴定到肽段序列
    """
    if diann_seq in pfind_seqs:
        identify_cnt += 1
        continue
    else:  # DIA-NN鉴定到而pFind可靠结果中未出现的target肽段
        not_identify_cnt += 1

    """
    pFind找到了该肽段序列，但没通过阈值
    """
    if diann_seq in pfind_seqs_bad:
        """
        判断DIA-NN肽段序列结果的可信程度
        """
        find_not_pass_cnt += 1
        if seq_conf[diann_seq] > 0:
            find_not_pass_conf_cnt += 1
        else:
            find_not_pass_unknown_cnt += 1
        continue

    """
    剩下的是pFind真的没找到的肽段序列：
        先验证DIA-NN鉴定到肽段的可信程度：如果在DIA-NN鉴定到了多种形式的相同肽段，则判其正确，此时需要考察pFind未鉴定到、
            或在相同谱图鉴定到错误肽段的原因；否则，其正确性未知，此时若pFind在相同谱图上鉴定到了不同肽段，则需考察其正确性
            (是否鉴定到多种形式的相同肽段)。
    """
    not_find_cnt += 1
    if seq_conf[diann_seq] > 0:  # DIA-NN鉴定到的肽段序列是可信的，但pFind却找到，这是需要重点关注的肽段
        not_find_conf_cnt += 1
        
        flag_mass_problem = False
        flag_identify_problem = False
        
        for pfind_idx in diann_seq2idxs[diann_seq]:  # 遍历DIA-NN肽段序列被鉴定的谱图
            # 先从这些"混合谱"中找到对应谱图
            diann_ress = idx2diann_res[pfind_idx]
            for diann_res in diann_ress:
                if diann_res[0] == diann_seq:
                    diann_modified_seq, diann_charge, diann_cal_mz = diann_res[1], diann_res[2], diann_res[4]
                    break

            # 看看pFind是否有解析出对应的母离子
            left_mz, right_mz = diann_cal_mz*(1-ppm_level*1e-6), diann_cal_mz*(1+ppm_level*1e-6)
            precursors_info = scan2precursors_info[pfind_idx]
            flag_find_mz = False
            flag_identify_problem = False
            for precursor_info in precursors_info:
                mz, charge = precursor_info
                if left_mz <= mz <= right_mz:  # 找到对应的质荷比范围
                    flag_find_mz = True

            if flag_find_mz:  # 只要有一个没找到母离子，则有mass问题
                focused_spectra.append((pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, ",".join(seq2info[diann_seq][1]), "mass"))
                flag_identify_problem = True
            else:  # 只要有一个pParse解析出了对应母离子，但结果不是对应肽段序列，则鉴定有问题
                focused_spectra.append((pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, ",".join(seq2info[diann_seq][1]), "identify"))
            
                flag_mass_problem = True

            del diann_modified_seq, diann_charge, diann_cal_mz  # 只是为了验证每轮都一定可以找到对应的diann_info

        if flag_mass_problem:
            not_find_conf_mass_cnt += 1
        if flag_identify_problem:
            not_find_conf_identify_cnt += 1
            
    else:  # DIA-NN鉴定到的肽段序列是可信程度不确定，且pFind却找到
        not_find_unknown_cnt += 1
        
        flag_mass_problem = False
        flag_identify_problem = False
        
        for pfind_idx in diann_seq2idxs[diann_seq]:  # 遍历DIA-NN肽段序列被鉴定的谱图
            # 先从这些"混合谱"中找到对应谱图
            diann_ress = idx2diann_res[pfind_idx]
            for diann_res in diann_ress:
                if diann_res[0] == diann_seq:
                    diann_modified_seq, diann_charge, diann_cal_mz = diann_res[1], diann_res[2], diann_res[4]
                    break

            # 看看pFind是否有解析出对应的母离子
            left_mz, right_mz = diann_cal_mz*(1-ppm_level*1e-6), diann_cal_mz*(1+ppm_level*1e-6)
            precursors_info = scan2precursors_info[pfind_idx]
            flag_find_mz = False
            flag_identify_problem = False
            for precursor_info in precursors_info:
                mz, charge = precursor_info
                if left_mz <= mz <= right_mz:  # 找到对应的质荷比范围
                    flag_find_mz = True

            if flag_find_mz:  # 只要有一个没找到母离子，则有mass问题
                focused_spectra_.append((pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, ",".join(seq2info[diann_seq][1]), "mass"))
                flag_identify_problem = True
            else:  # 只要有一个pParse解析出了对应母离子，但结果不是对应肽段序列，则鉴定有问题
                focused_spectra_.append((pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, ",".join(seq2info[diann_seq][1]), "identify"))
                flag_mass_problem = True

            del diann_modified_seq, diann_charge, diann_cal_mz  # 只是为了验证每轮都一定可以找到对应的diann_info

        if flag_mass_problem:
            not_find_unknown_mass_cnt += 1
        if flag_identify_problem:
            not_find_unknown_identify_cnt += 1


print('DIA-NN与pFind鉴定一致的肽段序列数目:', identify_cnt)
print('DIA-NN鉴定到而pFind未鉴定到的肽段序列数目:', not_identify_cnt)
print('-'*100)
print('DIA-NN鉴定到，而未通过pFind阈值的肽段序列数目:', find_not_pass_cnt)
print('\t其中可信肽段序列数目', find_not_pass_conf_cnt)
print('\t其中正确性未知的肽段序列数目', find_not_pass_unknown_cnt)
print('-'*100)
print('DIA-NN鉴定到而pFind未找到的肽段序列数目:', not_find_cnt)
print('\t其中可信肽段序列的个数:', not_find_conf_cnt)
# 该肽段对应的所有谱图中，存在未检出质荷比的情况
print('\t\tpFind未找到源于母离子质荷比未检测出的次数:', not_find_conf_mass_cnt)
# 该肽段对应的所有谱图中，存在检出质荷比但未鉴定到该肽段的情况
print('\t\tpFind未鉴定到源于鉴定问题的次数:', not_find_conf_identify_cnt)
print('\t其中正确性未知的肽段个数:', not_find_unknown_cnt)
# 该肽段对应的所有谱图中，存在未检出质荷比的情况
print('\t\tpFind未找到源于母离子质荷比未检测出的次数:', not_find_unknown_mass_cnt)
# 该肽段对应的所有谱图中，存在检出质荷比但未鉴定到该肽段的情况
print('\t\tpFind未找到源于鉴定问题的次数:', not_find_unknown_identify_cnt)
```

```python
"""
将重点关注的肽段序列对应的谱图写入文件：pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, protein_names, problem
"""
shortened_name = shorten_name(raw_paths[0])
focused_spectra_conf_path = f"data/{shortened_name.split('_')[0]}_focused_spectra_conf.txt"
with open(focused_spectra_conf_path, "w") as f:
    for spectra_info in focused_spectra:
        f.write("\t".join([str(info) for info in spectra_info])+'\n')
print(f"重点关注肽段序列(可信)数目为{not_find_conf_cnt}, 其对应{len(focused_spectra)}张谱图")

focused_spectra_unknown_path = f"data/{shortened_name.split('_')[0]}_focused_spectra_conf.txt"
with open(focused_spectra_unknown_path, "w") as f:
    for spectra_info in focused_spectra:
        f.write("\t".join([str(info) for info in spectra_info])+'\n')
print(f"重点关注肽段序列(正确性未知)数目为{not_find_unknown_cnt}, 其对应{len(focused_spectra_)}张谱图")
```

## 将重点关注肽段序列对应的谱图写入mgf文件中

```python
idx2focused_spectra = dict()
for spectrum_info in focused_spectra:
    pfind_idx, diann_seq, diann_modified_seq, diann_charge, diann_cal_mz, protein_names = spectrum_info
    if pfind_idx in idx2focused_spectra:
        idx2focused_spectra[pfind_idx].append((diann_seq, diann_modified_seq, diann_charge, diann_cal_mz))
    else:
        idx2focused_spectra[pfind_idx] = [(diann_seq, diann_modified_seq, diann_charge, diann_cal_mz)]
```

```python
"""
将重点关注的肽段序列对应的谱图构造raw文件:
    BEGIN IONS
    TITLE=20230108_AST_Neo1_DDA_UHG_HeLa_200ng_2th2p5ms_top50_01_20230808155512.196.196.2.0.dta
    CHARGE=2+
    RTINSECONDS=63.6365200
    PEPMASS=421.216677
    [MZ, INTENSITY]
    ...
    [MZ, INTENSITY]
    END IONS
"""
import random

focused_spectra_path = focused_spectra_conf_path
output_mgf_paths = [path.replace(".raw", "_conf.mgf") for path in raw_paths]


def dump(title_line, charge_line, rt_line, pepmass_line, mz_intensity_lines, fw):
    fw.write("BEGIN IONS\n")
    fw.write(title_line+charge_line+rt_line+pepmass_line+"".join(mz_intensity_lines))
    fw.write("END IONS\n")


ori_info_num = 0
cur_info_num = 0
ppm_random = 0  # 随机增加[-ppm_random,ppm_random]ppm的随机扰动
for ori_mgf_path, output_mgf_path in zip(mgf_paths, output_mgf_paths):
    print(ori_mgf_path, output_mgf_path)
    # break
    shortened_name = shorten_name(ori_mgf_path)

    fr = open(ori_mgf_path, "r")
    with open(output_mgf_path, "w") as fw:
        ms2_scan_no_last = None
        TITLE_lines = list()
        CHARGE_lines = list()
        RT_lines = list()
        PEPMASS_lines = list()
        mz_intensity_lines = list()  # 同一scan_no对应的相同(mz,intensity)行
        while True:
            line = fr.readline()
            if not line:  # End Of File
                break
                
            if line.startswith("B"):  # BEGIN
                if ori_info_num % 1000 == 0:
                    print(ori_info_num)
            elif line.startswith("T"):  # TITLE: xxx.scan_no.scan_no.charge.i.dta
                xxx, ms2_scan_no, _, charge, i, dta = line.strip().split('=')[1].split('.')
                if ms2_scan_no_last is not None and ms2_scan_no_last != ms2_scan_no:
                    num = 0
                    # 把原先mgf中的写进去
                    for i in range(len(TITLE_lines)):
                        # dump(TITLE_lines[i], CHARGE_lines[i], RT_lines[i], PEPMASS_lines[i], mz_intensity_lines, fw)
                        # cur_info_num += 1
                        ori_info_num += 1
                        num += 1
                    # 写入新的谱图
                    pfind_ms2_idx = shortened_name + "-" + ms2_scan_no_last
                    added = idx2focused_spectra.get(pfind_ms2_idx, [])
                    for seq, modified_seq, charge, cal_mz in added:
                        title = ".".join([xxx, ms2_scan_no_last, ms2_scan_no_last, str(charge), str(num), "dta"])
                        title_line = f"TITLE={title}\n"
                        charge_line = f"CHARGE={charge}+\n"
                        pepmass = cal_mz*(1+random.uniform(-1,1)*ppm_random*1e-6)  # 增加随机扰动
                        pepmass_line = f"PEPMASS={pepmass}\n"
                        dump(title_line, charge_line, RT_lines[0], pepmass_line, mz_intensity_lines, fw)
                        
                        cur_info_num += 1
                        num += 1
                        
                    TITLE_lines = [line]
                    CHARGE_lines = list()
                    RT_lines = list()
                    PEPMASS_lines = list()
                    mz_intensity_lines = list()  # 同一scan_no对应的相同(mz,intensity)行
                else:
                    TITLE_lines.append(line)
                ms2_scan_no_last = ms2_scan_no

            elif line.startswith("C"):  # CHARGE
                CHARGE_lines.append(line)
            elif line.startswith("R"):  # RTINSECONDS
                RT_lines.append(line)
            elif line.startswith("P"):  # PEPMASS
                PEPMASS_lines.append(line)
            else:  # mz_intensity_lines
                if len(mz_intensity_lines) == 0:
                    mz_intensity_lines.append(line)
                    while True:
                        line = fr.readline()
                        if line.startswith("E"):  # END
                            break
                        mz_intensity_lines.append(line)
                else:
                    while True:
                        line = fr.readline()
                        if line.startswith("E"):  # END
                            break

        # 尾部情况
        if ms2_scan_no_last is not None:
            num = 0
            # 把原先mgf中的写进去
            for i in range(len(TITLE_lines)):
                # dump(TITLE_lines[i], CHARGE_lines[i], RT_lines[i], PEPMASS_lines[i], mz_intensity_lines, fw)
                # cur_info_num += 1
                ori_info_num += 1
                num += 1
            pfind_ms2_idx = shortened_name + "-" + ms2_scan_no_last
            added = idx2focused_spectra.get(pfind_ms2_idx, [])
            for seq, modified_seq, charge, cal_mz in added:
                title = ".".join([xxx, ms2_scan_no_last, ms2_scan_no_last, str(charge), str(num), "dta"])
                title_line = f"TITLE={title}\n"
                charge_line = f"CHARGE={charge}+\n"
                pepmass = cal_mz*(1+random.uniform(-1,1)*ppm_random*1e-6)  # 增加随机扰动
                # pepmass = charge*cal_mz - (charge-1)*P_mass  # PEPMASS实际上是m/z
                pepmass_line = f"PEPMASS={pepmass}\n"
                dump(title_line, charge_line, RT_lines[0], pepmass_line, mz_intensity_lines, fw)
                
                cur_info_num += 1
                num += 1
    fr.close()

# 多个.mgf文件
print(f"原始.mgf文件包含{ori_info_num}个母离子信息")
print(f"当前.mgf文件包含{cur_info_num}个母离子信息")
```
