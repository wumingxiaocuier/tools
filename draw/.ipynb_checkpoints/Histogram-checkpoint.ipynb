```python
def get_peptides(res_path):
    # pFind.Filtered.spectra
    peptides = set()
    f = open(res_path,"r")
    next(f,None)
    for line in f:
        items = line.strip().split("\t")
        seq = items[5]
        mod = items[10]
        peptide = seq+"\t"+mod
        peptides.add(peptide)
    f.close()
    return peptides
```

```python
target_res_path = ""
other_res_path = []

infos = list()  # [[shared,gained,loss],[gained,...],...]
target_peptides = get_peptides(target_res_path)
for res_path in other_res_paths:
    peptides = get_peptides(res_path)
    shared = len(target_peptides&peptides)
    gained = len(target_peptides-peptides)
    loss = len(peptides-target_peptides)
    infos.append([shared,gained,loss])

print(infos)
```

```python
"""
图10: 修改细打分后selectpeak=5000，pepnum=10。
四簇直方图，每簇三个柱子，分别为修改粗打分前、修改粗打分后、修改细打分后，
粗打分进入top40、细打分进入top10、重打分排名为top1、通过FDR阈值的谱图数目堆叠图。
"""
import numpy as np
import matplotlib.pyplot as plt

# 示例数据：每个簇有3个柱子，每个柱子有3个堆叠部分
data = [
    [100,[95,10,1],[90,20,2]],
    [100,[95,10,1],[90,20,2]],
]
clusters = len(data)  # 有几组数据集
bars = len(data[0])   # 每个数据集有几组数据(默认第一个为target)
nums = 3  # 每组数据(第一组除外)有3个部分-shared,gained,loss
            
# 柱子的宽度和位置
bar_width = 0.4
cluster_positions = np.arange(clusters)*2  # 簇的位置，稍微间隔开
indexes = np.arange(bars)*bar_width


# 创建一个图形和子图
fig, ax = plt.subplots(figsize=(8,4))

# 颜色列表
colors = ["#9d454e", "#cf9197", "#bfad72", "#e5dabc",]
labels = ["Find target m/z in MS1", "not Find target m/z in MS1"]

# 绘制每个簇的堆叠柱子
for cluster_idx, cluster_data in enumerate(data):
    # 绘制该簇第一根柱子
    ax.bar(cluster_positions[cluster_idx], cluster_data[0], bar_width, 
                   label="shared" if cluster_idx==0 else None,
                   color=colors[0], edgecolor="black")
    # 绘制后续柱子(堆叠图)
    for stack_idx, stack_data in enumerate(cluster_data[1:]):
        shared,gained,loss = stack_data
        # shared
        ax.bar(cluster_positions[cluster_idx]+indexes[stack_idx+1], shared, bar_width, bottom=0,
                   color=colors[0], edgecolor="black")
        # gained
        ax.bar(cluster_positions[cluster_idx]+indexes[stack_idx+1], gained, bar_width, bottom=shared,
                   label="gained" if cluster_idx==0 and stack_idx==0 else None ,
                   color=colors[1], edgecolor="black")
        # loss
        ax.bar(cluster_positions[cluster_idx]+indexes[stack_idx+1], -loss, bar_width, bottom=0,
               label="loss" if cluster_idx==0 and stack_idx==0 else None,
               color=colors[2], edgecolor="black")

# 添加标签和标题
ax.set_xlabel('Experiments')
ax.set_ylabel('Number of Identified Peptides')
# ax.set_ylim([0,150])
ax.set_title('')
ax.set_xticks(cluster_positions)  # 只设置第一个簇的X轴刻度位置作为示例
# ax.set_xticklabels(["top50","top75","top100","cycle"])  # 设置X轴刻度标签
ax.legend(loc='upper left')  # 将图例放在图表外部以避免遮挡
# ax.legend(title='Cluster and Stack Info', loc='upper right')  # 将图例放在图表外部以避免遮挡
# ax.spines['top'].set_visible(False)
# ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle='--', linewidth=0.5)

# 显示图表
# plt.tight_layout()
plt.show()
```
